```python
from plotnine import (
    ggplot,
    aes,
    geom_boxplot,
    geom_errorbar,
    geom_sina,
    geom_violin,
    labs,
    theme,
)
from plotnine.composition import plot_layout
from plotnine.data import mpg
```

```python
p = ggplot(mpg, aes("factor(year)", "hwy")) + labs(x="year")

p1 = p + geom_boxplot()
p2 = p + geom_violin()
p3 = p + geom_sina()
p4 = p + geom_errorbar(stat="summary")

p1 | p2 | p3
```

```python
(p1 | p2 | p3) + plot_layout(widths=(1, 1, 2))
```

When the layout explicitly has more columns than items, an empty space is created at the end.

```python
(p1 | p2 | p3) + plot_layout(ncol=4)
```

```python
(p1 / p2 / p3)
```

```python
(p1 / p2 / p3) + plot_layout(heights=(3, 2, 1))
```

```python
(p1 / p2 / p3) + plot_layout(nrow=4)
```

```python
p1 + p2 + p3
```

```python
(p1 + p2 + p3) + plot_layout(ncol=2)
```

```python
((p1 + p2) / p3) + plot_layout(heights=(1, 2))
```

Subcompositions can have their own layouts.

```python
c1 = (p1 + p2) + plot_layout(widths=(2, 1))
(c1 / p3) + plot_layout(heights=(1, 2))
```

```python
(p1 + p2 + p3 + p4) + plot_layout(ncol=3)
```

```python
(p1 + p2 + p3 + p4) + plot_layout(nrow=3, byrow=False)
```
