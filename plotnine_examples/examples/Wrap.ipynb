```python
from plotnine import (
    ggplot,
    aes,
    geom_boxplot,
    geom_errorbar,
    geom_violin,
    geom_sina,
    labs,
)
from plotnine.composition import plot_layout, Wrap
from plotnine.data import mpg
```

```python
from plotnine import *
p = ggplot(mpg, aes("factor(year)", "hwy")) + labs(x="year")

p1 = p + geom_boxplot()
p2 = p + geom_violin()
p3 = p + geom_sina()
p4 = p + geom_errorbar(stat="summary")

p1 + p2 + p3 + p4
```

Which is equivalent to

```python
Wrap([p1, p2, p3])
```

As you add more plots, they they wrapped in a grid.

```python
p1 + p2 + p3 + p4
```

The default wrapping into rows and columns is similar to that of `facet_grid`.
And, you can set the number of rows and/or columns.

```python
(p1 + p2 + p3 + p4) + plot_layout(ncol=3)
```
